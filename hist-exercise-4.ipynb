```python
import pandas as pd

data = pd.read_csv('census_data.csv')
data.head()
```

```python
county_code = 370
data = data[data.county == county_code]
```

```python
years = [1820, 1830, 1840]
pops = []
for y in years:
    pop = data[data.year == y][['citypop']].sum()
    pops.append(pop[0])

df = pd.DataFrame.from_dict({'population' : pops, 'year' : years})
```

```python
df.plot.line(x='year', y='population', xticks = years)
```
